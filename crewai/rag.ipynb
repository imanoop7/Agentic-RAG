```python
from crewai_tools import PDFSearchTool
rag_tool = PDFSearchTool(pdf='word.pdf',
    config=dict(
        llm=dict(
            provider="ollama", # or google, openai, anthropic, llama2, ...
            config=dict(
                model="llama3.2",
            ),
        ),
        embedder=dict(
            provider="ollama", # or openai, ollama, ...
            config=dict(
                model="nomic-embed-text",
            ),
        ),
    )
)
```

```python
rag_tool.run("What is word2veec")
```
